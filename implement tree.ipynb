```python
"""Simple Tree data structure implemented from scratch"""


class Node:
    def __init__(self, value):
        self.value = value
        self.children = []

    def add(self, value):
        """Add a child and return it (allows chaining)."""
        child = Node(value)
        self.children.append(child)
        return child

    def preorder(self):
        return [self.value] + [v for c in self.children for v in c.preorder()]

    def height(self):
        return 1 + max((c.height() for c in self.children), default=-1)

    def size(self):
        return 1 + sum(c.size() for c in self.children)

    def display(self, level=0):
        print("    " * level + str(self.value))
        for c in self.children:
            c.display(level + 1)


```

```python

root = Node("Company")
eng = root.add("Engineering")
eng.add("Backend")
eng.add("Frontend")
root.add("Marketing").add("SEO")

```

```python
root.display()
print("Pre-order:", root.preorder())
print("Size:", root.size(), "| Height:", root.height())
```
